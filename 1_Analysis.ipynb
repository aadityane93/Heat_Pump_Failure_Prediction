```python
import sys
import pandas as pd

# 1. Add the repository's 'src' folder to the Python path so we can import the module
sys.path.append('/content/heapo/src')

# 2. Import the dataloader
from heapo import HEAPO

# 3. Define the path to the extracted dataset
data_path = '/content/heapo_data/'

# 4. Initialize the dataloader
# use_local_time=True converts UTC timestamps to Europe/Zurich time
print("Initializing HEAPO dataloader (this verifies the dataset structure)...")
heapo = HEAPO(data_path=data_path, use_local_time=True, suppress_warning=False)
print("Dataset loaded successfully!\n")

# 5. Explore the available data
# Get a list of all household IDs
all_households = heapo.get_all_households()
print(f"Total households available: {len(all_households)}")

# Get the first household ID as an example
first_household = all_households[0]
print(f"Example Household ID: {first_household}\n")

# 6. Load 15-minute smart meter data for this household
print(f"Loading 15-minute smart meter data for household {first_household}...")
smd_df = heapo.load_smart_meter_data(household_id=first_household, resolution='15min')

# Display the first few rows
print("\nFirst 5 rows of Smart Meter Data:")
display(smd_df.head())

# 7. Load corresponding weather data for this household
weather_df = heapo.load_weather_data(id=first_household, resolution='daily')
print("\nFirst 5 rows of Daily Weather Data:")
display(weather_df.head())
```
